```python
dbutils.widgets.text("year", "")
dbutils.widgets.text("month", "*")
dbutils.widgets.text("context", "")

year = dbutils.widgets.get("year")
month = dbutils.widgets.get("month")
context = dbutils.widgets.get("context")

if year == "" or context == "":
    raise Exception("year and context paramenters are mandatory.")
```

```python
raw_path = f"dbfs:/databricks-datasets/nyctaxi/tripdata/{context}/*_{year}-{month}.csv.gz"
```

```python
df_raw = (spark.read
 .format("csv")
 .option("header", "true")
 .load(raw_path))
```

```python
(df_raw.write
 .mode("overwrite")
 .option("overwriteSchema", "true")
 .saveAsTable("bronze_trip_data")
 )
```
